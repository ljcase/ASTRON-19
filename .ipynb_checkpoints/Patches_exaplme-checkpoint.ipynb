```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import patches
```

```python
f = plt.figure(figsize=(6,6))

ax = f.add_subplot(111,xticks=[],yticks=[])

ax.set_xlim([0,1])
ax.set_ylim([0,1])

rectangle = patches.Rectangle((0.05,0.05),0.4,0.4,edgecolor = 'none',facecolor='blue',alpha=0.5)
ax.add_patch(rectangle)

circle = patches.Circle((0.6,0.6),0.3,edgecolor='none',facecolor='red', alpha=0.5)
ax.add_patch(circle)

ellipse = patches.Ellipse((0.8,0.8),0.3,0.4,edgecolor='none',facecolor='green', alpha = 0.5)
ax.add_patch(ellipse)

```

```python
f = plt.figure(figsize=(6,6))

ax = f.add_subplot(111,xticks=[],yticks=[])

ax.set_xlim([0,1])
ax.set_ylim([0,1])

rectangle = patches.Rectangle((0.05,0.05),0.4,0.4,edgecolor = 'none',facecolor='blue',zorder=-2)
ax.add_patch(rectangle)

circle = patches.Circle((0.6,0.6),0.3,edgecolor='none',facecolor='red',zorder=2)
ax.add_patch(circle)

ellipse = patches.Ellipse((0.8,0.8),0.3,0.4,edgecolor='none',facecolor='green',zorder=1)
ax.add_patch(ellipse)

```
