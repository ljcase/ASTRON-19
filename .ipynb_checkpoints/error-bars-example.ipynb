```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.fabs(10) + 2*np.random.randn(10)
y = np.fabs(10) + 2*np.random.randn(10)
```

```python
xerr = 2*np.random.random(10)
yerr = 2*np.random.random(10)
```

```python
plt.errorbar(x,y,xerr=xerr,yerr=yerr,fmt="bo")
plt.savefig('error-bar-example.png' , bbox_inches = 'tight', dpi=600)
```
