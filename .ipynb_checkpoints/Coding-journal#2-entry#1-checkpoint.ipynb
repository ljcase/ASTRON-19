```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def main():
    x = np.arange(0, 1, 0.01)
    
    y = np.exp(x)

    print(y)
if __name__ =="__main__":
    main()
```

In a new cell use the function to set y=exp(x), and then plot x vs. y. Label the x-axis as “Time [milliseconds]” and the y-axis as “Awesomeness”. Save the figure as a PDF.  


```python
y = np.exp(x)
x = np.arange(0, 1, 0.01)

plt.plot(x, xlabel = 'Time [milliseconds]', color = 'red')
plt.plot(y, ylabel = 'Awesomeness', color = 'b')
plt.savefig('Code-Journal#2-Entry#1.pdf')
```
