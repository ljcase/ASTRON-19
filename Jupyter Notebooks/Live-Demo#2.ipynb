```python
import numpy as  np
import matplotlib.pyplot as plt
```

```python
npoints=100
x = np.linspace(0.,2*np.pi,npoints)
y = np.sin(5 * x)
```

```python
plt.plot(x, y)                     # x-values first, then y-values
plt.ylim([-1.1, 1.1])                # call the function, do NOT assign
plt.xlabel('x')                    # call the function, do NOT assign
plt.ylabel('y = sin(5x)')          # call the function
plt.savefig('Live-Demo#2.png', facecolor='green')  # save before show
plt.show()
```
