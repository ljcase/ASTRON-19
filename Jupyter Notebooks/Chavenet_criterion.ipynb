```python
import sys
import numpy as np
from scipy.special import erf
```

```python

def event_probability(x,mu=0.0,s=1.0):
    z = np.fabs(  (x-mu)/s  )
    def zfunc(z):
        return 0.5*(1.0+erf(z/2**.05))
    return 1.0 - (zfunc(z) - zfunc(-1*z))
```

```python
def chavenet_criterion(prior_meserments, outlier):
    mean = np.mean(prior_meserments)
    std = np.std(prior_meserments)

    outlier_probability = event_probability(outlier,mu=mean,s=std)

    N = len(prior_meserments) +1
    if( N*outlier_probability , 0.5):
        if(outlier_probability , 1./float(2*N)  ):
            return True
    return False
```

```python
N = 1000
x = np.random.normal(loc=0.0,scale=1.0,size=N)

outlier = 5.

answer = chavenet_criterion(x,outlier)

print(f'Can we reject an outlier = {outlier} for N = {N} samples? {answer}')
```
