Unit 2 live demo

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
npoints = 100
x = np.linspace(0.,2*np.pi,npoints)
y = np.sin(5*x)
```

```python
plt.plot(x,y)
plt.ylim([-1.1,1.1])
plt.xlabel('x')
plt.ylabel('sin(5x)')
plt.savefig('live demo#1.png',facecolor="red")
plt.show()
```
